```python
import matplotlib.pylab as plt
import numpy as np
```

```python
def one_iteration(x, r):
    return r * x * (1 - x)


def one_run(x_0, r, n_steps):
    resutls = [x_0]

    for i in range(n_steps):
        resutls.append(one_iteration(resutls[-1], r))
    return resutls
```

```python
n_steps = 50
plt.scatter(range(n_steps + 1), one_run(0.2, 3.5, n_steps))
```

### Bifurcations plot

```python
def bifurcatoins_plot(x_0=0.2, n_transient=5, n_steps=5_000, r_range=(2.4, 4), r_step=0.005):
    x = []
    y = []
    for r in np.linspace(r_range[0], r_range[1], int((r_range[1] - r_range[0]) / r_step)):
        trj = one_run(x_0, r, n_steps)
        x += [r] * (n_steps - n_transient)
        y += trj[(n_transient + 1):]
    
    return x, y

plt.subplots(figsize=(10, 7))
plt.scatter(*bifurcatoins_plot(n_transient=500), s=0.0001, color='k')
plt.show()
```
